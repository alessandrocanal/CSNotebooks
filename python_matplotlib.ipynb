In Matplotlib, the relationship between Figures, Axes (or Subplots), and the plt.plot() function comes down to how Matplotlib manages its state.

The Anatomy:
- Figure vs. Axes
Figure (`plt.Figure`): This is the overall window or page where everything is drawn. It is the blank canvas that holds one or more plots (subplots), titles, legends, etc.
- Axes / Subplot (`plt.Axes`): This is the actual "plot" or graph inside the figure. It contains the x-axis, y-axis, ticks, labels, and the plotted data (lines, scatters, etc.). A Figure can have many Axes, but an Axes can only belong to one Figure.

## Why `plt.plot()` works without calling `plt.figure()`

Because `pyplot` (the `plt` module) is a **stateful interface** (a state machine), modeled on MATLAB. It keeps track internally of a "current figure" and a "current axes", and automatically creates whatever is missing when needed.

### What happens when you call `plt.plot(...)`

1. `plt.plot` internally calls `plt.gca()` ("get current axes").
2. `gca()` calls `plt.gcf()` ("get current figure").
3. If no figure exists, `gcf()` creates a new one with `plt.figure()`.
4. If the figure has no axes, one is added (`add_subplot(111)`).
5. The plot is drawn on that axes.

So the figure is always there; matplotlib simply creates it for you, implicitly.

### Verification

```python
import matplotlib.pyplot as plt

print(plt.get_fignums())   # [] -> no figure yet
plt.plot(range(0, 55, 5))
print(plt.get_fignums())   # [1] -> created automatically
fig = plt.gcf()            # get the Figure object
ax = plt.gca()             # get the Axes object
```

### Two usage styles

- **pyplot style (implicit):** convenient for quick scripts and notebooks. All `plt.xxx()` calls act on the current figure/axes.
- **Object-oriented style (explicit):** clearer and more controllable, recommended when you have multiple figures or subplots:

```python
fig, ax = plt.subplots()
ax.plot(range(0, 55, 5))
```

With multiple figures, the implicit style can be confusing ("which one am I drawing on?"), because `plt` always acts on the last active figure. For this reason, the second style is preferable in more structured code.

## plt.subplots() in matplotlib
This function creates a **figure** and one or more **subplots** (axes objects) at once. It's a very common way to build more flexible and professional plots.

````python
x = [1, 2, 3]
y = [4, 5, 6]

fig = plt.figure()
ax = fig.add_subplot(1,1,1)
ax.plot(x, y)
ax.set_title("Simple Plot")
plt.show()
````

Let's create more *subplot*

````python
fig = plt.figure()
fig.set(figwidth=14, figheight=8)
ax1 = fig.add_subplot(2,2,1)
ax2 = fig.add_subplot(2,2,2)
ax1.plot(x, y)
ax1.set_title("Graph 1")

ax2.hist(np.random.standard_normal(100), bins=20, color="green", alpha=0.3)
ax2.set_title("Graph 2")
````

````python
# an alternative way to create a 2x2 grid
fig, ax = plt.subplots(2,2)
print(ax)
# as you see ax is a list so...
ax[0,0].plot(x, y)
ax[0,0].set_title("Graph 1")
ax[0,1].plot(x, [10, 8, 6])
ax[0,1].set_title("Graph 2")
````

How can I represent two trends on the same graph?

````python
x = [1, 2, 3]
y1 = [4, 5, 6]
y2 = [6, 5, 4]

fig = plt.figure()
ax = fig.add_subplot(1,1,1)

ax.plot(x, y1, label="Andamento A")
ax.plot(x, y2, label="Andamento B")

ax.set_title("Confronto tra due andamenti")
ax.set_xlabel("x")
ax.set_ylabel("valore")

ax.legend()  # set the legend
plt.show()
````

## Random walk
Let's create a random walk

````python
step_option = [1,-1]
step_choice = choice(step_option)

walk = []
walk.append(step_choice)
for step in range(1,1000):
    next_step = walk[step-1] + choice(step_option)
    walk.append(next_step)

plt.plot(walk)
plt.show()
````

We can refactor our code using functions

````python
def rand_walk(step_num):
    walk = []
    step_choice = choice([1,-1])
    walk.append(step_choice)
    for i in range(1,step_num):
        step_choice_2 = choice([1,-1])
        next_step = walk[i-1] + step_choice_2
        walk.append(next_step)
    return walk

def plot_walk(walk):
    plt.plot(walk)
    plt.xlabel('Number of Steps')
    plt.ylabel('Distance from Origin')
    plt.title('Our Random Walk')
    plt.show()

random_w = rand_walk(100)
plot_walk(rand_walk(1000))
````

And finally we can create a new function that can draw multiple random walks

````python
def plot_walks(num,walk_steps):

    lab_list = list(range(1,num+1))
    for i in range(0,num):

        x = list(range(1,walk_steps+1))

        plt.plot(x,rand_walk(walk_steps),label = 'Plot Number' + ' ' + str(lab_list[i]))
        plt.xlabel('Number of Steps')
        plt.ylabel('Distance from Origin')
        plt.title('Our Random Walk')
        plt.legend(loc='lower left')
    plt.show()

plot_walks(5,100000)
````

You can do also with *subplots* and *scatter*

````python
def sc_rand_walk(step_num):
    walk = []
    step_choice = choice([1,-1])
    walk.append(step_choice)
    for i in range(1,step_num):
        step_choice_2 = choice([1,-1])
        next_step = walk[i-1] + step_choice_2
        walk.append(next_step)
    return walk

def sc_plot_walks(num, walk_steps):
    fx, ax = plt.subplots()
    for i in range(1,num):
        y = sc_rand_walk(walk_steps)
        x = range(1, len(y)+1)

        ax.scatter(x,y, label = 'Plot Number' + ' ' + str(i), s=10)
        ax.set_xlabel('Number of  Steps')
        ax.set_ylabel('Distance from Origin')
        ax.set_title('Our Random Walk')
        ax.legend(loc='lower left')
    plt.show()

sc_plot_walks(5,1000)
````

## SIerpinski Triangle
This is a very nice app that is based upon 3 transformations:
First Transformation
\begin{equation}
x_{n+1} = 0.5x_n
\end{equation}
\begin{equation}
y_{n+1} = 0.5y_n
\end{equation}
Second Transformation
\begin{equation}
x_{n+1} = 0.5x_n + 0.5
\end{equation}
\begin{equation}
y_{n+1} = 0.5y_n + 0.5
\end{equation}
Third Transformation
\begin{equation}
x_{n+1} = 0.5x_n + 1
\end{equation}
\begin{equation}
y_{n+1} = 0.5y_n
\end{equation}

Let's develop a Python script that generate some points according to one trasformation at time choosen randoly


````python
def trans_1(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x
    y1 = 0.5 * y
    return x1,y1

def trans_2(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x + 0.5
    y1 = 0.5 * y + 0.5
    return x1,y1

def trans_3(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x + 1
    y1 = 0.5 * y
    return x1,y1


transformations = [trans_1,trans_2,trans_3]
a1 = [0]
b1 = [0]
a,b = 0,0

for i in range(10000):
    trans = choice(transformations)
    a,b = trans((a,b))
    a1.append(a)
    b1.append(b)
````

````python
plt.rc('figure',figsize=(16,16))

plt.scatter(a1,b1,s=5)
#plt.savefig('my_figure.png')
plt.show()
````
